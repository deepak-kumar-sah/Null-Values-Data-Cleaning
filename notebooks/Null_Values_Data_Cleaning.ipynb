```python
df_clean = df.dropna(how="all").copy()

df_clean.shape
```

```python
len(df) - len(df_clean)
```

```python
df_clean.isnull().sum()
```

```python
(df_clean.isnull().mean() * 100).round(2)
```

```python
numeric_cols = df_clean.select_dtypes(include="number").columns

categorical_cols = df_clean.select_dtypes(include="object").columns

numeric_cols,categorical_cols
```

```python
df_clean[numeric_cols].isnull().sum()
```

```python
df_clean[categorical_cols].isnull().sum()
```

```python
for col in numeric_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())
```

```python
df_clean[numeric_cols].isnull().sum()
```

```python
df_clean.isnull().sum().sum()
```

```python
for col in categorical_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])
```

```python
df_clean[categorical_cols].isnull().sum()
```

```python
df_clean.isnull().sum().sum()
```

```python
df.shape,df_clean.shape
```

```python
df_clean.dtypes
```

```python
integer_cols = [
    "Age",
    "Performance",
    "Projects",
    "Joining_Year"
]

for col in integer_cols:
    df_clean[col] = df_clean[col].astype("Int64")
```

```python
df_clean.dtypes
```

```python
df_clean["Employee_ID"].isnull().sum()
```

```python
(df_clean["Employee_ID"] % 1 ==0).all()
```

```python
df_clean["Employee_ID"] = df_clean["Employee_ID"].astype("Int64")
```

```python
df_clean.dtypes
```

```python
text_cols = [
    "Name",
    "Gender",
    "Department",
    "City",
    "Education"
]

for col in text_cols:
    df_clean[col] = df_clean[col].astype("string")
```

```python
df_clean.dtypes
```

```python
df_clean.to_excel("../data/cleaned_dataset.xlsx",index=False)

df_clean.to_csv("../data/cleaned_dataset.csv",index=False)
```
